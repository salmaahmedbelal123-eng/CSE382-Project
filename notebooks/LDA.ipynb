```python
import pandas as pd

df = pd.read_csv("../data/interim/cleaned.csv")

display(df.head())
print("Shape:", df.shape)
print("Columns:", df.columns.tolist())
```

```python
Y = df["IsCanceled"]

X = df.drop(columns=["IsCanceled"])
```

```python
# Removed the cols  (ReservationStatus, ReservationStatusDate) because the model could learn to look at these cols 
X = X.drop(
    columns=["ReservationStatus", "ReservationStatusDate"],
    errors="ignore"
)
display(X.head())
```

```python
print("X shape:", X.shape)
print("y shape:", Y.shape)

print("\nTarget distribution:")
print(Y.value_counts())

print("\nFeature columns:")
print(X.columns.tolist())
```

```python


numeric_columns = X.select_dtypes(include=["number"]).columns.tolist()
categorical_columns = X.select_dtypes(exclude=["number"]).columns.tolist()

print("Number of numerical columns:", len(numeric_columns))
print(numeric_columns)

print("\nNumber of categorical columns:", len(categorical_columns))
print(categorical_columns)
```

```python

for col in categorical_columns:
    print(f"\n{col}:")
    print("Number of unique values:", X[col].nunique())
    print(X[col].value_counts().head(10))
```

```python

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    Y,
    test_size=0.2,
    random_state=42,
    stratify=Y
)

print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
print("y_train shape:", y_train.shape)
print("y_test shape:", y_test.shape)
```

```python

from sklearn.preprocessing import OneHotEncoder

encoder = OneHotEncoder(
    handle_unknown="ignore",
    sparse_output=False
)

X_train_cat = encoder.fit_transform(
    X_train[categorical_columns]
)

X_test_cat = encoder.transform(
    X_test[categorical_columns]
)

print("Encoded training shape:", X_train_cat.shape)
print("Encoded testing shape:", X_test_cat.shape)
```

```python

from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_num = scaler.fit_transform(
    X_train[numeric_columns]
)

X_test_num = scaler.transform(
    X_test[numeric_columns]
)

print("Numerical training shape:", X_train_num.shape)
print("Numerical testing shape:", X_test_num.shape)
```

```python

import numpy as np

X_train_prepared = np.hstack([
    X_train_num,
    X_train_cat
])

X_test_prepared = np.hstack([
    X_test_num,
    X_test_cat
])

print("Prepared training shape:", X_train_prepared.shape)
print("Prepared testing shape:", X_test_prepared.shape)
```

```python

import numpy as np

print("NaN values in X_train_prepared:",
      np.isnan(X_train_prepared).sum())

print("NaN values in X_test_prepared:",
      np.isnan(X_test_prepared).sum())
```
